```python
from google.colab import auth
from google.cloud import bigquery
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# 1. Authentification (une fois par session Colab)
# ============================================================
auth.authenticate_user()
client = bigquery.Client(project="sous-nos-pieds")
```

```python
query = """
    SELECT
        date_swi_clean AS date,
        AVG(SWI_UNIF_MENS) AS swi_moyenne,
        MIN(SWI_UNIF_MENS) AS swi_min,
        MAX(SWI_UNIF_MENS) AS swi_max
    FROM `sous-nos-pieds.staging.swi_par_commune`
    WHERE date_swi_clean IS NOT NULL
    GROUP BY date_swi_clean
    ORDER BY date_swi_clean
"""

df = client.query(query).to_dataframe()
print(f"{len(df)} lignes récupérées, de {df['date'].min()} à {df['date'].max()}")
df.head()
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
import pandas as pd
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display
from IPython.display import display, clear_output
```

```python
import plotly.graph_objects as go

fig = go.Figure()

# Bande min/max (zone remplie)
fig.add_trace(go.Scatter(
    x=df["date"], y=df["swi_max"],
    line=dict(width=0),
    showlegend=False,
    name="Max",
    hovertemplate="Max : %{y:.3f}<extra></extra>"
))
fig.add_trace(go.Scatter(
    x=df["date"], y=df["swi_min"],
    fill="tonexty",
    fillcolor="rgba(70, 90, 220, 0.2)",
    line=dict(width=0),
    showlegend=False,
    name="Min",
    hovertemplate="Min : %{y:.3f}<extra></extra>"
))

# Ligne de moyenne
fig.add_trace(go.Scatter(
    x=df["date"], y=df["swi_moyenne"],
    line=dict(color="blue", width=1.5),
    name="Moyenne",
    hovertemplate="Moyenne : %{y:.3f}<extra></extra>"
))

fig.update_layout(
    title=f"Évolution du SWI (moyenne, min, max) - {df['date'].min().year}-{df['date'].max().year}",
    xaxis_title="Date",
    yaxis_title="SWI",
    template="plotly_white",
    hovermode="x unified"
)

fig.update_xaxes(
    rangeslider_visible=True,
    rangeselector=dict(
        buttons=list([
            dict(count=1, label="1 an", step="year", stepmode="backward"),
            dict(count=5, label="5 ans", step="year", stepmode="backward"),
            dict(count=10, label="10 ans", step="year", stepmode="backward"),
            dict(count=20, label="20 ans", step="year", stepmode="backward"),
            dict(step="all", label="Tout")
        ])
    )
)

fig.show()
```

```python
import plotly.graph_objects as go
import plotly.colors as pc
import numpy as np

fig = go.Figure()

# Bande min/max (zone remplie)
fig.add_trace(go.Scatter(
    x=df["date"], y=df["swi_max"],
    line=dict(width=0),
    showlegend=False,
    name="Max",
    hovertemplate="Max : %{y:.3f}<extra></extra>"
))
fig.add_trace(go.Scatter(
    x=df["date"], y=df["swi_min"],
    fill="tonexty",
    fillcolor="rgba(70, 90, 220, 0.2)",
    line=dict(width=0),
    showlegend=False,
    name="Min",
    hovertemplate="Min : %{y:.3f}<extra></extra>"
))

# --- Courbe de moyenne colorée selon un seuil ---
seuil = 0.5  # <-- à ajuster selon ta valeur
couleur_bas = "rgba(70, 90, 220, 1)"   # bleu, en dessous du seuil
couleur_haut = "rgba(220, 50, 50, 1)"  # rouge, au-dessus du seuil

n = len(df)
valeurs = df["swi_moyenne"].values

for i in range(n - 1):
    v_moy = (valeurs[i] + valeurs[i + 1]) / 2  # valeur moyenne du segment
    couleur = couleur_haut if v_moy > seuil else couleur_bas
    fig.add_trace(go.Scatter(
        x=df["date"].iloc[i:i+2],
        y=df["swi_moyenne"].iloc[i:i+2],
        mode="lines",
        line=dict(color=couleur, width=1.5),
        showlegend=False,
        hovertemplate="Moyenne : %{y:.3f}<extra></extra>"
    ))

fig.update_layout(
    title=f"Évolution du SWI (moyenne, min, max) - {df['date'].min().year}-{df['date'].max().year}",
    xaxis_title="Date",
    yaxis_title="SWI",
    template="plotly_white",
    hovermode="x unified"
)

fig.update_xaxes(
    rangeslider_visible=True,
    rangeselector=dict(
        buttons=list([
            dict(count=1, label="1 an", step="year", stepmode="backward"),
            dict(count=5, label="5 ans", step="year", stepmode="backward"),
            dict(count=10, label="10 ans", step="year", stepmode="backward"),
            dict(count=20, label="20 ans", step="year", stepmode="backward"),
            dict(step="all", label="Tout")
        ])
    )
)

fig.show()
```
